```python
# Assignment 10: Bellman Optimality and Value Iteration
# 1. Define states, actions and parameters
states = ["S0", "S1"]
all_states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

gamma = 0.9
theta = 0.000001

normal_reward = -1
wall_reward = -1
goal_reward = 10

# Grid:
# S0 | WALL
# S1 | G

# 2. Define deterministic transitions
# Invalid moves and wall collisions keep the agent
# in its current state and give reward -1.

transitions = {
    ("S0", "Up"):    ("S0", normal_reward),
    ("S0", "Down"):  ("S1", normal_reward),
    ("S0", "Left"):  ("S0", normal_reward),
    ("S0", "Right"): ("S0", wall_reward),

    ("S1", "Up"):    ("S0", normal_reward),
    ("S1", "Down"):  ("S1", normal_reward),
    ("S1", "Left"):  ("S1", normal_reward),
    ("S1", "Right"): ("G", goal_reward)
}

print("1. MDP DETAILS")
print("States:", all_states)
print("Actions:", actions)
print("Discount factor:", gamma)
print("Normal reward:", normal_reward)
print("Wall reward:", wall_reward)
print("Goal reward:", goal_reward)

# 3. Verify all transitions
print("\n2. TRANSITION MAPPING")
for state in states:
    for action in actions:
        next_state, reward = transitions[(state, action)]
        print(f"{state} --{action}--> {next_state}, "
              f"Reward = {reward}")

assert len(transitions) == len(states) * len(actions)
print("\nAll state-action pairs have defined transitions.")

# 4. Initialize state values
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

print("\n3. INITIAL STATE VALUES")
print(V)

# 5. Perform synchronous Value Iteration
print("\n4. VALUE ITERATION")
print(f"{'Iteration':<12}{'V(S0)':<15}"
      f"{'V(S1)':<15}{'Delta':<15}")

iteration = 0

while True:
    iteration += 1
    new_V = V.copy()
    delta = 0.0

    # Use only previous iteration values
    for state in states:
        q_values = []

        for action in actions:
            next_state, reward = transitions[(state, action)]
            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    for state in states:
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    print(f"{iteration:<12}{V['S0']:<15.6f}"
          f"{V['S1']:<15.6f}{delta:<15.6f}")

    if delta < theta:
        break

# 6. Display optimal state values
print("\n5. FINAL OPTIMAL STATE VALUES")
for state in all_states:
    print(f"V*({state}) = {V[state]:.6f}")

# 7. Calculate optimal action values Q*(s, a)
print("\n6. ACTION VALUES Q*(s, a)")
Q = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[(state, action)]
        Q[state][action] = reward + gamma * V[next_state]
        print(f"Q*({state}, {action}) = "
              f"{Q[state][action]:.6f}")

# 8. Extract optimal policy, including ties
print("\n7. OPTIMAL POLICY")

for state in states:
    best_value = max(Q[state].values())
    best_actions = [
        action for action, value in Q[state].items()
        if abs(value - best_value) < 1e-9
    ]
    print(f"Optimal action(s) for {state}: {best_actions}")

# 9. Verify Bellman Optimality
print("\n8. BELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(Q[state].values())
    verified = abs(V[state] - max_q) < 1e-6

    print(f"{state}: V* = {V[state]:.6f}, "
          f"max Q* = {max_q:.6f}, "
          f"Verified = {verified}")

# 10. Explain the result
print("\n9. RESULT")
print("The optimal action from S0 is Down.")
print("The optimal action from S1 is Right.")
print("The agent reaches the goal from S1 and receives +10.")
print("The discount factor 0.9 reduces the value of future rewards.")
print("Bellman Optimality verification completed.")
```
