```python
# Assignment 11: Policy Iteration in Deterministic Grid World

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

gamma = 0.9
theta = 0.000001

# Transition mapping: (next state, reward)
transitions = {
    ("S0", "Up"):    ("S0", -1),
    ("S0", "Down"):  ("S1", -1),
    ("S0", "Left"):  ("S0", -1),
    ("S0", "Right"): ("S0", -1),

    ("S1", "Up"):    ("S0", -1),
    ("S1", "Down"):  ("S1", -1),
    ("S1", "Left"):  ("S1", -1),
    ("S1", "Right"): ("G", 100)
}

# Initialize policy and state values
policy = {"S0": "Up", "S1": "Up"}
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

print("POLICY ITERATION")
print("Discount factor:", gamma)
print("Initial policy:", policy)

# Display transitions
print("\nTRANSITIONS")
for state in states:
    for action in actions:
        next_state, reward = transitions[(state, action)]
        print(f"{state} --{action}--> {next_state}, Reward={reward}")

iteration = 0

while True:
    iteration += 1
    print(f"\nITERATION {iteration}")
    old_policy = policy.copy()

    # Policy Evaluation
    while True:
        delta = 0.0
        new_V = V.copy()

        for state in states:
            next_state, reward = transitions[(state, policy[state])]
            new_V[state] = reward + gamma * V[next_state]
            delta = max(delta, abs(new_V[state] - V[state]))

        V = new_V

        if delta < theta:
            break

    print("State values:", {
        s: round(V[s], 6) for s in ["S0", "S1", "G"]
    })

    # Calculate action values and improve policy
    Q = {}

    for state in states:
        Q[state] = {}
        for action in actions:
            next_state, reward = transitions[(state, action)]
            Q[state][action] = reward + gamma * V[next_state]

        best_action = max(Q[state], key=Q[state].get)
        policy[state] = best_action

    print("Action values:")
    for state in states:
        print(state, {
            a: round(q, 6) for a, q in Q[state].items()
        })

    print("Updated policy:", policy)

    # Stop when policy becomes stable
    if policy == old_policy:
        break

# Final results
print("\nFINAL OPTIMAL VALUES")
for state in ["S0", "S1", "G"]:
    print(f"V*({state}) = {V[state]:.6f}")

print("\nOPTIMAL POLICY")
for state in states:
    print(f"{state} -> {policy[state]}")

print("\nBELLMAN OPTIMALITY VERIFICATION")
for state in states:
    next_q = {}
    for action in actions:
        next_state, reward = transitions[(state, action)]
        next_q[action] = reward + gamma * V[next_state]

    max_q = max(next_q.values())
    print(
        f"{state}: V*={V[state]:.6f}, "
        f"max Q={max_q:.6f}, "
        f"Verified={abs(V[state] - max_q) < 1e-5}"
    )

print("\nOptimal path: S0 -> S1 -> G")
```
