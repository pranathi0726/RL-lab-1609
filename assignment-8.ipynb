```python
# Discount factor
gamma = 0.9

# Rewards
normal_reward = -1
goal_reward = 10

# Policy
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

# Grid:
# S0  S1  G
# X   S3  S4

# Function to calculate state value using the path
def state_value_function(start_state):
    path = []
    rewards = []

    state = start_state

    while state != "G":

        path.append(state)

        # Follow the given policy
        action = policy[state]

        # State transitions
        if state == "S0" and action == "Right":
            next_state = "S1"
            reward = normal_reward

        elif state == "S1" and action == "Down":
            next_state = "S3"
            reward = normal_reward

        elif state == "S3" and action == "Right":
            next_state = "S4"
            reward = normal_reward

        elif state == "S4" and action == "Up":
            next_state = "G"
            reward = goal_reward

        rewards.append(reward)
        state = next_state

    path.append("G")

    # Calculate discounted return
    value = 0

    for i in range(len(rewards)):
        value += (gamma ** i) * rewards[i]

    print("Path:", " -> ".join(path))
    print("Rewards:", rewards)
    print("Discounted Return:", value)

    return value


# Function to calculate action value
def action_value_function(state, action):

    rewards = []
    path = [state]

    # First action is given by the question
    if state == "S0" and action == "Up":

        # Up from S0 is outside the grid.
        # Agent stays in S0.
        rewards.append(normal_reward)
        next_state = "S0"

        path.append("S0")

    else:
        print("This example is for S0 and Up.")
        return

    # After the first action, follow the policy
    while next_state != "G":

        state = next_state
        policy_action = policy[state]

        if state == "S0" and policy_action == "Right":
            next_state = "S1"
            reward = normal_reward

        elif state == "S1" and policy_action == "Down":
            next_state = "S3"
            reward = normal_reward

        elif state == "S3" and policy_action == "Right":
            next_state = "S4"
            reward = normal_reward

        elif state == "S4" and policy_action == "Up":
            next_state = "G"
            reward = goal_reward

        rewards.append(reward)
        path.append(next_state)

    # Calculate discounted return
    value = 0

    for i in range(len(rewards)):
        value += (gamma ** i) * rewards[i]

    print("\nAction:", action)
    print("Path:", " -> ".join(path))
    print("Rewards:", rewards)
    print("Action Value Q(S0, Up):", value)

    return value


# --------------------------------------------------
# Calculate State Value for S0
# --------------------------------------------------

print("STATE VALUE")
print("-------------------------")

V_S0 = state_value_function("S0")


# --------------------------------------------------
# Calculate Action Value for S0 and Up
# --------------------------------------------------

print("\nACTION VALUE")
print("-------------------------")

Q_S0_Up = action_value_function("S0", "Up")
```
